```python
# get mark of a student and give a grade based on the marks
mark = int(input("Enter the mark: "))
if mark >= 90:
    grade = "A+"
elif mark >= 80:
    grade = "A"
elif mark >= 70:
    grade = "B"
elif mark >= 60:
    grade = "C"
else:
    grade = "F"
print("The grade is:", grade)
```
